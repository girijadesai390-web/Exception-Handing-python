2. Boolean Filtering-
Using the same marks array, create a new array containing only students who scored 60 or more.

```python
import numpy as np
marks=np.array([65,78,92,56,88,74,95,61])
```

```python
marks=np.array([65,78,92,56,88,74,95,61])
new_array=marks[marks>60]
print(new_array)
```
